```python
#wap to read a number form the user and check it is prime or not using user defined function

n=int(input("enter the number"))
def prime(n):
  c=0
  for i in range(1,n+1):
    if n%i==0:
      c=c+1

  if c==2:
    return "prime"
  else:
    return "not prime"

prime(n)
```

```python
#recurssion : a function calling itself

#wap to find factorial of a number using recurssion
n=int(input("enter teh number: "))

def fac(n):
  if n==1:
    return 1

  else:
    return n*fac(n-1)

fac(n)

```

```python
#ctreat a class of rectangles and create meathods to display its area and perimeter


class Rectangle():
  def get_data(self):
    self.length=int(input("enter the length: "))
    self.bredth=int(input("enter the bredth"))
  def display_data(self):
    print(self.length)
    print(self.bredth)

  def area(self):
    print(self.length*self.bredth)

  def perimeter(self):
    print(2*(self.length+self.bredth))

r1=Rectangle()
r1.get_data()
r1.display_data()
r1.area()
r1.perimeter()
```

```python
#create a class of cirlces and print area and perimeter using meathods


class Circle():
  def get_data(self):
    self.r=int(input("Enter the radious"))

  def display_data(self):
    print(self.r)

  def area(self):
    print(3.14*self.r*self.r)

  def perimeter(self):
    print(2*3.14*self.r)

c1=Circle()

c1.get_data()
c1.display_data()
c1.area()
c1.perimeter()

```

```python
class Addition():
  def __init__(self,n):
    self.value=n
  def __add__(self,other):
    return self.value+other.value+10


a1=Addition(10)

a2=Addition(12)

print(a1+a2)
```
